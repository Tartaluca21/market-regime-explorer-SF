# 04 · Pipeline e app / Pipeline and app

**Input · IT:** Le funzioni dei notebook 01–03, `settings.json` e il CSV unico.

**Output · IT:** `python run.py`, risultati riproducibili, bundle e app Streamlit che carica file salvati.

**Input · EN:** Functions from Notebooks 01–03, `settings.json` and the single CSV.

**Output · EN:** `python run.py`, reproducible results, a model bundle and a Streamlit app that loads saved files.

**Prima di iniziare / Before starting:** [serializzazione / persistence](https://scikit-learn.org/stable/model_persistence.html), [cache Streamlit / Streamlit caching](https://docs.streamlit.io/develop/concepts/architecture/caching); contratti fra moduli e controlli per data / module contracts and dated checks.

**Avvio / Start:** Jupyter dalla cartella del progetto; import di `core` dalla radice con `settings.json` / Jupyter from the project folder; import `core` from the root containing `settings.json`.


## A · Pipeline e controlli / Pipeline and checks

### 1. Validare i passaggi / Validate the stages

**IT:** Controllate percorsi, finestre, split, seed, massimo 40 tentativi HMM e feature selezionate dopo `select_features`. Verificate date, colonne e ordine ad ogni passaggio, accettando la lista iniziale vuota.

**EN:** Check paths, windows, splits, seeds, the 40-attempt HMM limit and selected features after `select_features`. Check dates, columns and order at each stage, accepting an initially empty feature list.


### 2. Salvare il bundle / Save the bundle

**IT:** Salvate `results/model_bundle.joblib` con estimatori, scaler, feature ordinate, settings, confini, hash sorgente, versioni e timing. Ricaricatelo e verificate label e probabilità identiche su una finestra datata.

**EN:** Save `results/model_bundle.joblib` with estimators, scaler, ordered features, settings, boundaries, source hash, versions and timing. Reload it and check identical labels and probabilities on a dated window.


### 3. Collegare la pipeline / Connect the pipeline

**IT:** Implementate `pipeline.run_pipeline` in `core/pipeline.py`, avviata da `python run.py`, includendo proiezioni di stato in `results/regime_results.csv`. Se il confronto di rischio è attivo, esportate `results/forecast_comparison.csv`; in caso di fallimento impedite il caricamento di risultati obsoleti.

**EN:** Implement `pipeline.run_pipeline` in `core/pipeline.py`, launched by `python run.py`, including state projections in `results/regime_results.csv`. If the risk comparison is enabled, export `results/forecast_comparison.csv`; on failure, prevent stale results from being loaded.


### 4. Controllare l’integrazione / Check integration

**IT:** Create test in `tests/` per dati, finestre causali, training-only, ordine degli input, budget e continuità del filtro. Se attivate il rischio, verificate target nei confini, soglia training, purging e indipendenza dal test.

**EN:** Create tests in `tests/` for data, causal windows, training-only fitting, input order, the fit budget and filter continuity. If risk forecasting is enabled, check targets within boundaries, the training threshold, purging and test independence.


### 5. Verificare la ricostruzione / Verify rebuilding

**IT:** Eseguite pipeline e test da un processo pulito e controllate bundle e CSV. Riproducete una finestra datata e simulate un errore: il risultato deve essere indipendente da variabili del notebook e file precedenti.

**EN:** Run the pipeline and tests from a clean process and check the bundle and CSVs. Reproduce a dated window and simulate a failure: results must be independent of notebook variables and earlier files.


## B · App e consegna / App and delivery

### 6. Collegare Streamlit / Connect Streamlit

**IT:** Implementate `app.py` caricando bundle e CSV da `results/`, con cache e messaggi per file assenti o incompatibili. Verificate che cambiare un controllo non avvii fitting e che l’app resti utilizzabile senza l’esperimento opzionale.

**EN:** Implement `app.py` to load the bundle and CSVs from `results/`, with caching and messages for missing or incompatible files. Check that changing a control never triggers fitting and that the app remains usable without the optional experiment.


### 7. Aggiungere i controlli / Add controls

**IT:** Collegate intervallo di date, preset storici, split e scelta del modello ai grafici condivisi. Verificate periodi disponibili, finestre vuote o brevi e date identiche fra grafici e tabelle.

**EN:** Connect the date range, historical presets, split and model choice to the shared charts. Check available periods, empty or short windows and identical dates across charts and tables.


### 8. Mostrare risultati e confronto / Show results and comparison

**IT:** Mostrate profili, probabilità, proiezioni πₜ @ A, entropia e diagnostiche, distinguendo fitting, selezione, filtering e smoothing. Se esiste `results/forecast_comparison.csv`, mostrate Brier, average precision, prevalenza e conteggi per modello, split e campionamento.

**EN:** Show profiles, probabilities, πₜ @ A projections, entropy and diagnostics, distinguishing fitting, selection, filtering and smoothing. If `results/forecast_comparison.csv` exists, show Brier, average precision, prevalence and counts by model, split and sampling scheme.


### 9. Esportare le figure / Export the figures

**IT:** Implementate `plots.export_figures` in `results/figures/`, riutilizzando le figure dell’app. Controllate titoli, periodo, cutoff del campione, unità, leggibilità e file non vuoti.

**EN:** Implement `plots.export_figures` into `results/figures/`, reusing the app figures. Check titles, period, sample cutoff, units, readability and nonempty files.


### 10. Verificare la consegna / Check delivery

**IT:** Provate installazione, pipeline, test e controlli dell’app in un ambiente pulito, aggiornando il README con comandi verificati. Controllate che ogni schermata usi i risultati salvati e resti coerente quando l’esperimento di rischio è disattivato.

**EN:** Try installation, the pipeline, tests and app controls in a clean environment, updating the README with verified commands. Check that every view uses saved results and remains consistent when the risk experiment is disabled.
